```python
import spacy
import os
import json
import torch

# from bertopic import BERTopic
from collections import defaultdict
from tqdm import tqdm

spacy_model = "en_core_web_trf"
spacy.require_gpu()
nlp = spacy.load(spacy_model)
```

```python
from typing import Iterator
from spacy.tokens import Doc


def batcher(data, batch_size):
	for idx in range(0, len(data), batch_size):
		yield data[idx : idx + batch_size]


def process_texts(texts, batch_size=16):
	processed_texts: Iterator[Doc] = nlp.pipe(texts, batch_size=batch_size, n_process=1)
	info = []

	for text in processed_texts:
		info += [
			(
				[ent.text for ent in text.ents],
				[ent.label_ for ent in text.ents],
				len([tok for tok in text if tok.is_punct is False]),
			)
		]
	return info
```

# Topic Modelling: LIAR-RAW + RAWFC

```python
import pickle
from collections import defaultdict
import os
import json
import torch
from bertopic import BERTopic

# from umap import UMAP
# from hdbscan import HDBSCAN
from sklearn.feature_extraction.text import CountVectorizer
from bertopic.representation import MaximalMarginalRelevance
from cuml.cluster import HDBSCAN
from cuml.manifold import UMAP
from collections import defaultdict
from tqdm import tqdm
from bertopic.representation import KeyBERTInspired
from sentence_transformers import SentenceTransformer


def load_rawfc_data(split):
	input_path_dir = f"../datasets/data/RAWFC/{split}"
	data = [
		json.load(open(f"{input_path_dir}/{json_file}", "r"))
		for json_file in os.listdir(input_path_dir)
	]
	sents = [
		sent["sent"]
		for sample in data
		for report in sample["reports"]
		for sent in report["tokenized"]
	]
	return list(set(sents))


def load_liaraw_data(split):
	input_path_dir = f"../datasets/data/LIAR-RAW/{split}.json"
	data = json.load(open(input_path_dir, "r"))
	sents = [
		sent["sent"]
		for sample in data
		for report in sample["reports"]
		for sent in report["tokenized"]
	]
	return list(set(sents))


sentence_model = SentenceTransformer("all-MiniLM-L6-v2")


def create_bertopic(size_data):
	umap_model = UMAP(n_neighbors=50, n_components=10, metric="cosine", random_state=42, min_dist=0)
	hdbscan_model = HDBSCAN(
		min_samples=10,
		min_cluster_size=int(size_data * 0.01),
		metric="euclidean",
		prediction_data=True,
		cluster_selection_method="eom",
	)
	vectorizer_model = CountVectorizer(stop_words="english", ngram_range=(1, 2))
	keybert_model = KeyBERTInspired(top_n_words=50, nr_repr_docs=10)
	mmr_model = MaximalMarginalRelevance(top_n_words=5, diversity=0.1)
	representation_model = [keybert_model, mmr_model]
	topic_model = BERTopic(
		verbose=True,
		nr_topics="auto",
		embedding_model=sentence_model,
		umap_model=umap_model,
		hdbscan_model=hdbscan_model,
		vectorizer_model=vectorizer_model,
		representation_model=representation_model,
	)
	return topic_model


dataset_names = ["RAWFC", "LIAR-RAW"]
splits = ["train", "test", "val"]
BATCH_SIZE = 64
for dataset in dataset_names[:]:
	for split in splits[:]:
		output_path = f"topics/{dataset}_{split}"
		os.makedirs(output_path, exist_ok=True)
		print(dataset, split)
		# if "representative_docs.json" in os.listdir(output_path):
		# 	continue
		match dataset:
			case "RAWFC":
				load_data_func = load_rawfc_data
			case "LIAR-RAW":
				load_data_func = load_liaraw_data
		sents = load_data_func(split=split)

		topic_model = create_bertopic(size_data=len(sents))

		if "embds.pkl" not in os.listdir(output_path):
			embds = sentence_model.encode(
				sents, show_progress_bar=True, batch_size=BATCH_SIZE, normalize_embeddings=True
			)
			pickle.dump(embds, open(os.path.join(output_path, "embds.pkl"), "wb"))
		else:
			embds = pickle.load(open(os.path.join(output_path, "embds.pkl"), "rb"))
		topics, probs = topic_model.fit_transform(sents, embeddings=embds)
		if topic_model._outliers == 1:
			print("================= Outliers! =================")
			topics = topic_model.reduce_outliers(
				sents, topics, strategy="embeddings", embeddings=embds, threshold=0.75
			)
			topics = topic_model.reduce_outliers(
				sents, topics, strategy="probabilities", probabilities=probs
			)
			topic_model.update_topics(
				sents,
				topics=topics,
				vectorizer_model=topic_model.vectorizer_model,
				representation_model=topic_model.representation_model,
			)
		topic_labels = topic_model.generate_topic_labels(
			nr_words=3, topic_prefix=False, word_length=None, separator=", "
		)
		topic_model.set_topic_labels(topic_labels)
		topic_model.save(
			output_path, serialization="safetensors", save_ctfidf=True, save_embedding_model=True
		)
		output_representative_docs = topic_model.representative_docs_

		sent2topic_mapping = {
			sent: topic_idx for sent, topic_idx in zip(sents, topic_model.topics_)
		}
		sent2emb_mapping = {sent: emb for sent, emb in zip(sents, embds)}
		json.dump(
			output_representative_docs,
			open(os.path.join(output_path, "representative_docs.json"), "w"),
			indent=2,
		)
		pickle.dump(sent2emb_mapping, open(os.path.join(output_path, "sent2emb_mapping.pkl"), "wb"))
		json.dump(
			sent2topic_mapping,
			open(os.path.join(output_path, "sent2topic_mapping.json"), "w"),
			indent=2,
		)
```

```python
import json
import os


def load_entities(path):
	files = os.listdir(path)
	files = sorted(files, key=lambda x: int(x.split(".")[0]))
	data = [json.load(open(os.path.join(path, file), "r")) for file in files]
	return data


datasets = ["RAWFC", "LIAR-RAW"]
splits = ["train", "val", "test"]
da = datasets[0]
sp = splits[1]
dir_input = f"spacy_processed/{da}_{sp}"
```

```python
from collections import defaultdict

ents_evidence = load_entities(f"{dir_input}")
ents_claim = load_entities(f"{dir_input}_claim")
ents_explain = load_entities(f"{dir_input}_explain")
files_names = os.listdir(dir_input)
files_names = sorted(files_names, key=lambda x: int(x.split(".")[0]))

reports_not_all_coverage = defaultdict(list)
for evidence, claim, explain, file in zip(ents_evidence, ents_claim, ents_explain, files_names):
	event_id = file.removesuffix(".json")

	claim.pop("claim")
	claim.pop("veracity_label")
	sample_ents_claim = list(claim.values())[0][0][0]
	sample_ents_claim = set(sample_ents_claim)

	explain.pop("explain")
	explain.pop("veracity_label")
	sample_ents_explain = list(explain.values())[0][0][0]
	sample_ents_explain = set(sample_ents_explain)

	evidence.pop("veracity_label")
	sample_reports_evidence = list(evidence.values())[0]
	for report_idx, report in enumerate(sample_reports_evidence):
		report = list(report.values())[0]
		ents = set()
		for sentence in report:
			ents |= set(sentence[1])
		num_shared_ents_evidence_claim = len(ents & sample_ents_claim)
		ratio_missing_ents_claim = (
			num_shared_ents_evidence_claim / len(sample_ents_claim)
			if len(sample_ents_claim) > 0
			else 0
		)
		num_shared_ents_evidence_explain = len(ents & sample_ents_explain)
		ratio_missing_ents_explain = (
			num_shared_ents_evidence_explain / len(sample_ents_explain)
			if len(sample_ents_explain) > 0
			else 0
		)

		if len(sample_ents_claim) < 1:
			ratio_missing_ents_claim = -1
		if len(sample_ents_explain) < 1:
			ratio_missing_ents_explain = -1

		claim_ents_missing = sample_ents_claim - ents
		explain_ents_missing = sample_ents_explain - ents

		# print(ratio_missing_ents_claim)
		# print(ratio_missing_ents_explain)
		reports_not_all_coverage[event_id] += [
			{
				"report_idx": report_idx,
				"evidence_ents": ents,
				"claim": (ratio_missing_ents_claim, claim_ents_missing, sample_ents_claim),
				"explain": (ratio_missing_ents_explain, explain_ents_missing, sample_ents_explain),
			}
		]
```

```python
ratios_claim = []
ratios_explain = []
for event_id, sample in reports_not_all_coverage.items():
	ratios_claim += [[report["claim"][0] for report in sample]]
	ratios_explain += [[report["explain"][0] for report in sample]]
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import itertools

ratios_claim_flat = list(itertools.chain.from_iterable(ratios_claim))
ratios_explain_flat = list(itertools.chain.from_iterable(ratios_explain))

sns.set_theme(style="whitegrid")

sns.violinplot(
	{
		"claim": ratios_claim_flat,
		"explain": ratios_explain_flat,
	},
	inner="quart",
)
plt.ylabel("Ratio of entities from evidence\nthat are present in target text")
plt.title("Across all reports")

sns.despine()
plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import itertools


def get_quarts(ratios):
	stats = defaultdict(list)
	for report in ratios:
		if len(report) < 1:
			continue
		quart1, median, quart3 = np.quantile(report, [0.25, 0.5, 0.75])
		stats["median"] += [median]
		stats["quart1"] += [quart1]
		stats["quart3"] += [quart3]
		stats["max"] += [np.max(report)]
		stats["min"] += [np.min(report)]
	return stats


quarts_per_sample_claim = get_quarts(ratios_claim)
quarts_per_sample_explain = get_quarts(ratios_explain)
```

```python
sns.set_theme(style="whitegrid")

sns.violinplot(quarts_per_sample_claim, inner="quart", cut=0)
plt.ylabel("Ratio of entities from evidence\nthat are present in target text")
plt.title("Evidence-Claim: Across all reports")

sns.despine()
plt.show()
```
